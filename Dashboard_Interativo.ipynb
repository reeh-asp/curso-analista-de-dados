```python
import plotly.express as px
import pandas as pd
from dash import Dash, html, dcc
from dash.dependencies import Input, Output
```

```python
def cria_graficos(df):

    fig1 = px.bar(df, x='Gênero', y='Qtd_Vendidos', title='Quantidade de Vendas por Gênero',color='Gênero')

    fig2 = px.pie(df, names='Gênero', color='Gênero', hole=0.4, title='Representatividade dos Gêneros sobre a Quantidade de Vendas')

    df['Qtd_Vendidos'] = pd.to_numeric(df['Qtd_Vendidos'], errors='coerce')
    df['Qtd_Vendidos'] = df['Qtd_Vendidos'].fillna(0)
    fig3 = px.scatter(df, x='Temporada', y='Marca', size='Qtd_Vendidos', color='Gênero' )
    fig3.update_layout(title='Quantidade de Vendas por Marca e Temporada')
    
    return fig1, fig2, fig3
```

```python
path = r'C:\Users\resil\Desktop\ANALISTA DE DADOS\Nova pasta\ecommerce_estatistica.csv'

df = pd.read_csv(path)

Lista_Gênero = df['Gênero'].unique()
options = [{'label':gênero,'value':gênero} for gênero in Lista_Gênero] 
```

```python
def cria_app(df):
    app = Dash(__name__)
      
    app.layout = html.Div([
        html.H1('Dashboard Interativo'),
        html.Br(),
        html.H2('Analise da influencia de Cada Gênero nas Vendas'),
        dcc.Checklist(
            id='id_gênero',
            options=options,
            value=[Lista_Gênero[0]],
    ),
        dcc.Graph(id='fig1'),
        dcc.Graph(id='fig2'),
        dcc.Graph(id='fig3'),
    ])
    return app



if __name__ == '__main__':
    
    app = cria_app(df)
       
    @app.callback(
        [
            Output('fig1','figure'),
            Output('fig2','figure'),
            Output('fig3','figure'),        
        ],
        [Input('id_gênero', 'value')]
    )
    def atualiza_grafico(Lista_Gênero):
        filtrar_df = df[df['Gênero'].isin(Lista_Gênero)]
        fig1, fig2, fig3 = cria_graficos(filtrar_df) 
        return fig1, fig2, fig3
    
    app.run_server(debug=True, port=8051)
```
